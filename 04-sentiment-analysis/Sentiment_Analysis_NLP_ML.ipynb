```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt', sep= ';', header = None, names = ['text', 'emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions = df['emotion'].unique()
```

```python
emotion_numbers = {}

i = 0
for emo in unique_emotions:
    emotion_numbers[emo] = i
    i += 1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df
```

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
import string

def remove_punc(text):
  return text.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
df
```

```python

def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

```python
df
```

```python

def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emojis)
```

```python
import nltk
```

```python
df
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.head()
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)

```

```python
df['text'] =  df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfTransformer
```

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
```

```python
X_train_bow = bow_vectorizer.transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB()
```

```python
nb_model.fit(X_train_bow, y_train)
```

```python
pred_nb = nb_model.predict(X_test_bow)
```

```python
print(accuracy_score(y_test, pred_nb))
```

```python
tfidf_vectorizer = TfidfTransformer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train_bow)
X_test_tfidf = tfidf_vectorizer.transform(X_test_bow)

nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf, y_train)


```

```python
pred_nb2 = nb2_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, pred_nb2))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression()
```

```python
logistic_model.fit(X_train_tfidf, y_train)
```

```python
pred_logistic = logistic_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, pred_logistic))
```
